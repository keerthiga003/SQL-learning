```python
%%sql
CREATE DATABASE IF NOT EXISTS datetime_vs_timestamp;
USE datetime_vs_timestamp;
```

```python
%%sql
CREATE TABLE demo_events (
    event_id INT AUTO_INCREMENT PRIMARY KEY,
    event_name VARCHAR(100),
    event_date DATETIME,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

```python
%%sql
INSERT INTO demo_events (event_name, event_date)
VALUES
    ('New Year Celebration', '2025-01-01 00:00:00'),
    ('Summer Fest', '2025-06-15 12:30:00');
```

```python
%%sql
select * from demo_events
```

1.extract the year,date,day from the event_name column

```python
%%sql
SELECT 
    event_id,
    event_name,
    year(event_date) as event_year, 
    created_at 
FROM demo_events;
```

```python
%%sql
SELECT 
    event_id,
    event_name,
    date(event_date) as event_date, 
    created_at 
FROM demo_events;
```

```python
%%sql
SELECT 
    event_id,
    event_name,
    day(event_date) as event_day, 
    created_at 
FROM demo_events;
```

```python
#Add 30 days to the existing date

%%sql
SELECT 
    event_id,
    event_name,
    event_date , 
    event_date + interval 30 day as date_plus_30
FROM demo_events;
```

```python
#Formating

%%sql
SELECT event_id,event_name,DATE_FORMAT(event_date , '%W %M %e %Y') as formatted_date from demo_events ;
```

```python
%%sql
SELECT event_id,event_name,DATE_FORMAT(event_date , '%h:%i %p ') as formatted_date from demo_events ;
```
