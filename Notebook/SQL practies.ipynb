DDL-(DATA DEFINITION LANGUAGE)
CREATE - to create new records
ALTER - used for modify the existing database object 
DROP - to delet the object
TRUNCATE - to remove all the records from the table
RENAME - to rename the object
--------------------------
DML - DATA MANIPULATION LANGUAGE
SELECT - select the data or query data 
INSERT - to add new records
UPDATE - to modify the existing record
DELETE - toremove the records
--------------------------
DCL - DATA CONTROL LANGUAGE 
GRANT - to give the user access privilages
REVOKE - to take away privilages
--------------------------
TCL - TRANSACTION CONTROL LANGUAGE
COMMIT - to save transaction permanently 
ROLLBACK - to undo transaction
SAVEPOINT - to set a point within a transaction to which you can rollback
--------------------------

soft delete
---------
-> it means that the record is present in the table but it shows like it deleted
-> it is work if we are not using the "on casecade delete" because if we using the on casecade there is no problem in deletion of the record in both the rider and driver table

```python
%%sql
drop table if exists rides;
drop table if exists driver;
create table driver(
    driver_id int primary key,
    driver_name varchar(100),
    licence_number varchar(100) unique
);
create table rides(
    ride_id INT PRIMARY KEY,
    driver_id INT NOT NULL,
    rider_id INT NOT  NULL,
    pickup_location VARCHAR(100) NOT NULL,
    dropoff_location  VARCHAR(100) NOT NULL,
    ride_date DATETIME NOT NULL,
    fare DECIMAL(10,2) NOT NULL,
    foreign key (driver_id) references driver(driver_id));
```

minimal
--------
-> It means, we are create a composite key  (ie more than one primary key ->primary key(std_id,course_id)) ,if we delete the one  primary key that reduce  the uniqueness then it is called "Minimal"
----------------

Not minimal
---------
-> It means, we are create a composite key  (ie more than one primary key ->primary key(std_id,course_id,course_name)) ,if we delete the one  primary key (ie. course_name) that can not reduce  the uniqueness then it is called "Not Minimal"
