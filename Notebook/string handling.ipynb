```python
%%sql

CREATE TABLE CustomerDetails (
    id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    email VARCHAR(100),
    city VARCHAR(100),
    phone_number VARCHAR(15)
   
);
```

```python
%%sql
select customer_name,length(customer_name) as name_length
from CustomerDetails
```

```python
%%sql
select customer_name,length(customer_name) as name_length,
upper(city) as city_upper,
lower(customer_name) as cust_low,
concat (customer_name, ' : ',city," : TN") as name_c,
substring(customer_name,1,6) as name_prefix,
trim("   chennai   ") as trimed_city,  #it is used for remove the space
lpad(customer_name,15,"*") as left_padded_name # it is used when the customer_name length is less than 15 then it add the give symbol


from CustomerDetails
```

```python
%%sql
select replace(customer_name," ",'_') as updated_name,
instr(customer_name,'a') as position_of_a,#it return the index of the given character
left(customer_name,5) as first_5_chars,
right(customer_name,5) as last_5_chars,
reverse(customer_name) as reversed_name,
format(354354367,2) as formatted_number

from CustomerDetails
```

```python
%%sql
select 
upper(concat (customer_name, ' : ',city," : TN")) as name_c, #function within another function
length(upper(concat (customer_name, ' : ',city," : TN"))) as name_len
from CustomerDetails
```
