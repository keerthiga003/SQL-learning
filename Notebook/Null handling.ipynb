coalesce
----
It is used for replace the actual null value with the given value
ex: ->none to 0.00

```python
%%sql
select id,
    customer_name,
    amount,
    coalesce (amount , '0.00')
from CustomerData
```

ifnull
-----

```python
%%sql
select id,
    customer_name,
    amount,
    ifnull (amount , '0.00')
from CustomerData
```
