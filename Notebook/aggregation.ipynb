count
----

```python
%%sql
select count(*) from CustomerTransactions where transaction_status = "Failed"
```

```python
%%sql
select count(*) from CustomerTransactions where transaction_status = "Failed" and product = "Laptop"
```

```python
%%sql
select count(*) from CustomerTransactions where transaction_status in("Failed","Pending")
```

```python
%%sql
select count(*) from CustomerTransactions where transaction_status not in("Failed","Pending")
```

sum
---

```python
%%sql
select sum(amount) from CustomerTransactions;
```

```python
%%sql
select sum(amount) as total_transaction from CustomerTransactions where transaction_status="Completed"
```

min
---

```python
%%sql
select min(amount) as minimum_transaction from CustomerTransactions 
```

max
-----

```python
%%sql
select max(amount) as maximum_transaction from CustomerTransactions 
```

avg
----

```python
%%sql
select avg(amount) as average_transaction from CustomerTransactions 
```

Group by
------

```python
%%sql
select login_device,
sum(amount) as total_revenue
from CustomerTransactions 
group by login_device
```

Having
----

```python
%%sql
select login_device,
sum(amount) as total_revenue
from CustomerTransactions 
group by login_device
having
 sum(amount)>80000
```

```python
%%sql

CREATE TABLE CustomerData (
    id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    email VARCHAR(100),
    phone_number VARCHAR(15),
    address VARCHAR(200),
    amount DECIMAL(10, 2)
);
```

```python
%%sql
select 
   customer_name,
   amount,
   case  
      when amount > 4000 then "High spender"
      when amount between 2000 and 4000 then "Medium spender"
      when amount <= 2000 then "Low spender"
    else 'no data'
    end as spending
from CustomerData
```

```python
%%sql
select 
customer_name,
case
   when phone_number is null then email
   else phone_number 
end as contact
from CustomerData
```

```python
%%sql
select 
customer_name,
case
   when phone_number is null and email is null then '9999'
   when phone_number is null then email
   else phone_number 
end as contact
from CustomerData
```
